# **PySpark DataFrame Basics**


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("DataFrame Basics") \
    .master("local[*]") \
    .getOrCreate()
```

## Create a DataFrame

```python
data = [("Ravi",22,"Python"),
    ("Priya",23,"SQL"),
    ("Rahul",21,"Power BI"),
    ("Anu",24,"Python")]
```

```python
columns = ["Name","Age","Course"]
```

```python
df = spark.createDataFrame(data, columns)
df.show()
```

## Display Schema

```python
df.printSchema()
```

## Select Columns

```python
df.select("Name","Course").show()
```

## Filter Data

```python
df.filter(df.Age > 22).show()
```

## Group By

```python
df.groupBy("Course").count().show()
```

## Stop Spark

```python
spark.stop()
```
