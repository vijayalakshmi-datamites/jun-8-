# **PySpark RDD Basics**

Create Spark Session

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("RDD Basics") \
    .master("local[*]") \
    .getOrCreate()

sc = spark.sparkContext
```

Create an RDD

```python
numbers = [10, 20, 30, 40, 50]

rdd = sc.parallelize(numbers)

rdd.collect()
```

Number of Partitions

```python
rdd.getNumPartitions()
```

```python
square = rdd.map(lambda x: x*x)

square.collect()
```

```python
even = rdd.filter(lambda x: x % 20 == 0)

even.collect()
```

Actions

```python
print("Count :", rdd.count())
print("First :", rdd.first())
print("Take :", rdd.take(3))
```

```python
words = sc.parallelize(["Big", "Data", "Big", "Spark", "Data"])

word_count = (words
              .map(lambda word: (word,1))
              .reduceByKey(lambda a,b: a+b))

word_count.collect()
```

Stop Spark

```python
spark.stop()
```
